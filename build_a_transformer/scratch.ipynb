```python
import conf, dataset, embedding
import torch
import torch.nn as nn

import matplotlib.pyplot as plt
```

```python
model = torch.nn.Embedding(num_embeddings=conf.VOCAB_SIZE, embedding_dim=conf.EMBEDDING_DIM)
```

```python
x = torch.tensor([1, 2], dtype=torch.int32)
emb = model.forward(x)
```

```python
emb[0].dot(emb[1])
norms = []
for i in range(conf.VOCAB_SIZE):
    for j in range(i + 1, conf.VOCAB_SIZE):
        emb = model.forward(torch.tensor([i, j], dtype=torch.int))
        norm = torch.linalg.norm(emb[0] - emb[1])
        norms.append(norm.item())
```

```python
model.forward(torch.tensor([1, 1], dtype=torch.int32))
```
